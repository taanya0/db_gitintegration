```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, when

# Create DataFrame
data = [
    ("C001", 500),
    ("C002", 1500),
    ("C003", 3000)
]

df = spark.createDataFrame(data, ["customer_id", "amount"])

# If-Else condition
df_new = df.withColumn(
    "customer_type",
    when(col("amount") >= 2000, "Premium")
    .when(col("amount") >= 1000, "Gold")
    .otherwise("Regular")
)

df_new.show()
```
